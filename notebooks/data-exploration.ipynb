## Cross Check the relationship between three files

```python
# How many beneficiary IDs in disbursements DON'T exist in beneficiaries.csv?
orphan_ids = set(disbursements["beneficiary_id"]) - set(beneficiaries["beneficiary_id"])
print("Orphan beneficiary IDs in disbursements:", len(orphan_ids))
```

```python
# How many beneficiaries have NO matching national ID record?
matched_ids = set(national_ids["linked_beneficiary_id"].dropna())
unmatched = set(beneficiaries["beneficiary_id"]) - matched_ids
print("Beneficiaries with no national ID match:", len(unmatched))
```

```python
# Do CNIC formats match exactly between beneficiaries.csv and national_id_records.csv?
merged_check = beneficiaries.merge(
    national_ids, left_on="beneficiary_id", right_on="linked_beneficiary_id", how="inner"
)
mismatch_count = (merged_check["cnic"] != merged_check["national_id_no"]).sum()
print("CNIC mismatches between the two systems:", mismatch_count)
```

```python
missing_cnic = set(beneficiaries["cnic"]) - set(national_ids["national_id_no"])
print("CNICs in beneficiaries.csv but not in national_id_records.csv:", len(missing_cnic))
print("Missing CNICs:", missing_cnic)
```

```python
# Remove dashes, spaces, and convert to string to preserve leading zeros
beneficiaries['cnic_clean'] = beneficiaries['cnic'].astype(str).str.replace(r'[^0-9]', '', regex=True)
national_ids['cnic_clean'] = national_ids['national_id_no'].astype(str).str.replace(r'[^0-9]', '', regex=True)
```

```python
# 1. Raw Mismatches (with formatting like dashes)
raw_mismatches = set(beneficiaries['cnic']) - set(national_ids['national_id_no'])

# 2. Clean Mismatches (pure numbers only)
clean_mismatches = set(beneficiaries['cnic_clean']) - set(national_ids['cnic_clean'])

# 3. The Diagnosis
formatting_failures = len(raw_mismatches) - len(clean_mismatches)
true_missing = len(clean_mismatches)

print(f"Format-related mismatches (fixed by cleaning): {formatting_failures}")
print(f"True mismatches (CNIC completely missing): {true_missing}")
```

```python
# Find structurally broken CNICs in beneficiaries
beneficiaries['digit_count'] = beneficiaries['cnic_clean'].str.len()

# Filter for rows that are not exactly 13 digits long
invalid_length_beneficiaries = beneficiaries[beneficiaries['digit_count'] != 13]

print("Rows with invalid CNIC lengths (should be 13 digits):")
print(invalid_length_beneficiaries[['cnic', 'digit_count']])
```

```python
# 1. Merge the systems on beneficiary ID
merged_check = beneficiaries.merge(
    national_ids, left_on="beneficiary_id", right_on="linked_beneficiary_id", how="inner"
)

# 2. Filter for rows where the CNIC numbers do not match
mismatches_df = merged_check[merged_check["cnic"] != merged_check["national_id_no"]]

# 3. Create a clean report showing the differences using the correct pandas default suffixes (_x, _y)
cnic_mismatch_report = mismatches_df[[
    'beneficiary_id', 
    'full_name_x',       # Name recorded in beneficiaries registry
    'full_name_y',       # Name recorded at NADRA national database
    'cnic',              # CNIC recorded in welfare program
    'national_id_no'     # Actual CNIC recorded at NADRA
]].copy()

# Rename columns for perfect readability
cnic_mismatch_report.columns = ['ID', 'Welfare Name', 'NADRA Name', 'Welfare CNIC', 'NADRA CNIC']

print(f"Total Discrepancies Found: {len(cnic_mismatch_report)}")
cnic_mismatch_report.head(50)
```
