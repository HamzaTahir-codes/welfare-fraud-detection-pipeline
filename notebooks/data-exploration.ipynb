## Cross Check the relationship between three files

```python
# How many beneficiary IDs in disbursements DON'T exist in beneficiaries.csv?
orphan_ids = set(disbursements["beneficiary_id"]) - set(beneficiaries["beneficiary_id"])
print("Orphan beneficiary IDs in disbursements:", len(orphan_ids))
```

```python
# How many beneficiaries have NO matching national ID record?
matched_ids = set(national_ids["linked_beneficiary_id"].dropna())
unmatched = set(beneficiaries["beneficiary_id"]) - matched_ids
print("Beneficiaries with no national ID match:", len(unmatched))
```

```python
# Do CNIC formats match exactly between beneficiaries.csv and national_id_records.csv?
merged_check = beneficiaries.merge(
    national_ids, left_on="beneficiary_id", right_on="linked_beneficiary_id", how="inner"
)
mismatch_count = (merged_check["cnic"] != merged_check["national_id_no"]).sum()
print("CNIC mismatches between the two systems:", mismatch_count)
```
